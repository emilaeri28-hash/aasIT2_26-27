```python
høyde = [100, 120, 125, 155, 160]

for i in høyde:
    if i<115:
        print(i,"cm, du er dessverre for liten og kan ikke kjøre")
        continue
    elif i<125:
        print(i,"cm, du må kjøre med en voksen")
        continue

    print(i,"cm og kan kjøre alene")
```
